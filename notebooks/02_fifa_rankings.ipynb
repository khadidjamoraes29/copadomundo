Ranking da fifa de 1992 até 2024

```python
import pandas as pd
```

```python
ranking = pd.read_csv(
    "../data/fifa_ranking-2024-06-20.csv"
)
```

```python
ranking["rank_date"] = pd.to_datetime(
    ranking["rank_date"]
)
```

```python
ranking = ranking[
    [
        "country_full",
        "rank",
        "total_points",
        "rank_date",
        "confederation"
    ]
]
```

```python
TEAM_MAPPING = {
    "USA": "United States",
    "Korea Republic": "South Korea",
    "Korea DPR": "North Korea",
    "IR Iran": "Iran",
    "Türkiye": "Turkey"
}
```

```python
ranking["country_full"] = ranking[
    "country_full"
].replace(TEAM_MAPPING)
```

```python
ranking.to_csv(
    "../data/clean_rankings.csv",
    index=False
)
```
