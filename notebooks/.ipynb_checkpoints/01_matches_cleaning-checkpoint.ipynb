Notebook sobre partidas desde 1930 até 2022

```python
import pandas as pd
import numpy as np
```

```python
matches = pd.read_csv(
    "../data/matches_1930_2022.csv"
)
```

```python
selected_columns = [
    "home_team",
    "away_team",
    "Attendance",
    "Round",
    "Host",
    "Year",
    "Date",
    "home_score",
    "away_score"
]

matches = matches[selected_columns].copy()
```

```python
def get_result(row):

    if row["home_score"] > row["away_score"]:
        return "home_win"

    elif row["home_score"] < row["away_score"]:
        return "away_win"

    return "draw"


matches["result"] = matches.apply(
    get_result,
    axis=1
)
```

```python
matches["Date"] = pd.to_datetime(
    matches["Date"]
)
```

```python
TEAM_MAPPING = {
    "USA": "United States",
    "Korea Republic": "South Korea",
    "Korea DPR": "North Korea",
    "IR Iran": "Iran",
    "Türkiye": "Turkey"
}
```

```python
matches["home_team"] = matches[
    "home_team"
].replace(TEAM_MAPPING)

matches["away_team"] = matches[
    "away_team"
].replace(TEAM_MAPPING)
```

```python
matches.to_csv(
    "../data/clean_matches.csv",
    index=False
)
```
